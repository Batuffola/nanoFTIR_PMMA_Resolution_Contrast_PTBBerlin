# PMMA 800µm vs. 600µm – Resolution Comparison

_This notebook explores nano-FTIR measurements of PMMA deposited on Pt/SiO₂, examining spectral quality, SNR behavior, and contrast effects based on scan resolution and acquisition parameters._

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.fftpack import fft, fftfreq, ifft
from scipy.optimize import curve_fit
```

```python
def einlesen(textdatei):
    df = pd.read_csv(textdatei, sep = '\t', skiprows = 29, header = 0, engine= 'python')
    data = df[['Wavenumber', 'O2A', 'O2P','Omega']].to_numpy()
    #A= df['02A']
    #P= df['02p']
    return data
```

```python
data=einlesen('2020-06-23 174954 NF S Deep3_PMMA_on_Pt_BK_152mA.txt')
#print(einlesen('2020-06-23 174954 NF S Deep3_PMMA_on_Pt_BK_152mA.txt'))

n=0
P = np.zeros((1199, 1)) 
A= np.zeros((1199, 1))
W=np.zeros((1199, 1))
for i in range (n,n+1199):
    P [i-n]= data[i,2]
    A[i-n]= data[i,1]
    W[i-n]= data [i,0]

plt.plot(W, A, 'b-', label='O2A')
plt.plot (W,P,'r-', label='O2P')
plt.axis([1600, 1845, -0.2, 1])
plt.legend()
plt.xlabel('Wavenumbers')
plt.ylabel('Amplitude')
#Spektren (Amplitude und Phase) plotten, am besten jeweils immer in die 1730 und 1150 Region zoomen.
plt.show() 
plt.plot(W, A, 'b-', label='O2A')
plt.plot (W,P,'r-', label='O2P')
plt.legend()
plt.axis([1000, 1300, -0, 1.2])
#600 Punkte 50 ms Integrationszeit sowie über 20 Messungen gemittelt - Messzeit: ca. 10 min
# kein 1730 peak sichtbar, trotz langer Integration
plt.xlabel('Wavenumbers')
plt.ylabel('Amplitude')
#Spektren (Amplitude und Phase) plotten, am besten jeweils immer in die 1730 und 1150 Region zoomen.
plt.show() 
```
